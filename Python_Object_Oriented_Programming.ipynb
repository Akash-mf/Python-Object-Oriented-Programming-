```python
#4
class Cricket:
    def __init__(self, style, nation):
        self.style = style
        self.nation = nation

    def hitter(self):
        print(f"He is a {self.style}-hand batsman coming from {self.nation} and is a hitter.")

    def classic(self):
        print(f"He is a {self.style}-hand batsman coming from {self.nation} and is a classic player.")

    def batsman(self):
        print(f"He is a {self.style}-hand batsman coming from {self.nation} and is a batsman.")

# Creating objects
arun = Cricket("Right", "India")
mega = Cricket("Left", "Australia")
balaji = Cricket("Right", "India")

# Calling methods
mega.classic()
balaji.batsman()
arun.hitter()
```

```python
#7
class Ball:
    def __init__(self, colour, brand):
        self.colour = colour
        self.brand = brand

    def circle(self):
        print(f"It is circle in shape and colour {self.colour} and brand is {self.brand}")

    def rectangle(self):
        print(f"It is rectangle in shape and colour {self.colour} and brand is {self.brand}")  # ✅ Fixed "slef"

    def cube(self):
        print(f"It is cube in shape and colour {self.colour} and brand is {self.brand}")  # ✅ Fixed "slef"


# Child class - Basketball
class Basketball(Ball):
    def moving(self):
        print("The basketball can bounce")


# Child class - Football
class Football(Ball):
    def distance(self):
        print("The football can move a long distance")


# Creating object for Basketball
obj_basketball = Basketball("mickeymouse", "nike")

# Calling methods
obj_basketball.circle()
obj_basketball.moving()

# Creating object for Football
obj_football = Football("red", "adidas")
obj_football.rectangle()
obj_football.distance()
```

```python
#Abstraction
from abc import ABC,abstractmethod
class selva(ABC):
    def land1(self):
        pass
    def land2(self):
        pass
    def land3(self):
        pass
class asset1(selva):
    def land1(self):
        print("the land1 belongs to me")
class asset2(selva):
    def land2(self):
        print("the land2 belongs to me")
class asset3(selva):
    def land3(self):
        print("the land3 belongs to me")

mega=asset1()
arun=asset2()
balaji=asset3()

mega.land1()
arun.land2()
balaji.land3()
```

```python
#super function method
class father():
    def profo(self):
        print(" i am a pilot")
    def education(self):
        print(" it is a aeronaticaul")
    def age(self):
        print(" he is 45")
class child1(father):
    def number(self):
        super().age()
        print( " i am the first child")

#outside the class
mega=child1()

mega.number()
```

```python
# Encapsulation Example
# public, protected, private

class Kumar:
    def __init__(self, name="sla", address="chennai", income=500000):  # ✅ Fixed __init__
        self.name = name          # Public variable
        self._address = address   # Protected variable (by convention, single underscore)
        self.__income = income    # Private variable (double underscore)

    # Public method to access private variable
    def incomes(self):
        return self.__income

# Creating an object
abi = Kumar()

# Accessing public variable directly
print("Public:", abi.name)

# Accessing protected variable (possible, but not recommended)
print("Protected:", abi._address)

# Accessing private variable directly (will cause error if we try like abi.__income)
# print(abi.__income)  # ❌ This will raise an AttributeError

# Correct way to access private variable
print("Private (through method):", abi.incomes())
```

```python
#Encapsulation
class Employee:
    def __init__(self, name, salary):  # ✅ Fixed __init__ method
        self.name = name              # Public attribute
        self._department = "IT"       # Protected attribute (single underscore)
        self.__salary = salary        # Private attribute (double underscore)

    # Public method to display details
    def show_details(self):
        print(f"Name: {self.name}")
        print(f"Department: {self._department}")
        print(f"Salary: {self.__salary}")

    # Getter method for private salary
    def get_salary(self):
        return self.__salary


# ✅ Object created outside the class
emp = Employee("Akash", 50000)

# Accessing public variable directly
print(emp.name)            # Public → Allowed

# Accessing protected variable directly (Allowed, but not recommended)
print(emp._department)

# Accessing private variable correctly using getter
print(emp.get_salary())

# ❌ Wrong: Accessing private variable directly by original name (will cause AttributeError)
# print(emp.__salary)

# ✅ Right: Accessing private variable using name mangling (not recommended)
print(emp._Employee__salary)  # Just to demonstrate
```
